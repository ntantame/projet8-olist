```python
GRANDE_REGION = {
    'Norte':        ['AC', 'AP', 'AM', 'PA', 'RO', 'RR', 'TO'],
    'Nordeste':     ['AL', 'BA', 'CE', 'MA', 'PB', 'PE', 'PI', 'RN', 'SE'],
    'Centro-Oeste': ['DF', 'GO', 'MT', 'MS'],
    'Sudeste':      ['ES', 'MG', 'RJ', 'SP'],
    'Sul':          ['PR', 'RS', 'SC'],
}

ETAT_VERS_REGION = {
    etat: region
    for region, etats in GRANDE_REGION.items()
    for etat in etats
}

assert len(ETAT_VERS_REGION) == 27
print(" 27 états couverts")
```

```python
# Adapte le chemin si tes CSV sont ailleurs
customers = pd.read_csv('../data/raw/olist_customers_dataset.csv')
sellers = pd.read_csv('../data/raw/olist_sellers_dataset.csv')

# Vérification cruciale du brief
assert set(customers['customer_state']) <= set(ETAT_VERS_REGION)
assert set(sellers['seller_state']) <= set(ETAT_VERS_REGION)
print("Tous les états des CSV sont couverts par notre dictionnaire")
```
