## CONSTRUCTION DE LA DIMENSION PRDUIT

```python
# importation des bibliotheques
import  pandas as pd
from pathlib import Path
```

```python
# chargement des deux ficihiers csv necessaires
data_dir=Path('../data/raw')
produit=pd.read_csv(data_dir/'olist_products_dataset.csv')
translation=pd.read_csv(data_dir/'product_category_name_translation.csv')
print("total observation de produit:",len(produit))
print("total observation de translation:",len(translation))
```

```python
# construction du dataframe
dim_produit=produit.merge(
    translation,
    on='product_category_name',
    how='left'
).rename(columns={
        "product_category_name":"categorie_portugais",
        "product_category_name_english":"categorie_anglais",
        "product_weight_g":"poids",
    })[["product_id","categorie_portugais","categorie_anglais","poids"]]

print("total observation de produit:",len(dim_produit))
print("colonnes de la dimension: ", dim_produit.columns)
```

```python
# verification

# nombre total de ligne
assert len(dim_produit)==32951, f"resultat attendu: 32951, resultat obtenu: {len(dim_produit)}"
# abscence d edoublon sur product_id
assert dim_produit["product_id"].is_unique, " produit dupliques"

# verification des valeurs manquantes
print("dimension produit:")
print("total observation de produit:", len(dim_produit))
print("produit unique :",dim_produit["product_id"].nunique())
```

```python
# sauvegarde dans un fichier CSV
Path("../data/processed").mkdir(parents=True, exist_ok=True)
dim_produit.to_csv("../data/processed/dim_produit.csv", index=False, encoding='utf-8')
print("fichier enregistrer avec succes")
```
