## CREATION DE LA DIMENSION REGION

```python
# importation des bibliotheques
import  pandas as pd
```

```python
# creation  des 5 grandes region du bresil
grande_region={
    "Norte":["AC","AP","AM","PA","RO","RR","TO"],
    "Nordeste":["AL","BA","CE","MA","PB","PE","PI","RN","SE"],
    "Centro-Oeste":["DF","GO","MT","MS"],
    "Sudeste":["ES","MG","RJ","SP"],
    "Sul": ["PR","RS","SC"],
}

# inversement du dictionnaire etats - region
etat_region={e: r for r, etats in grande_region.items() for e in etats}

# construction du dataframe
dim_region=pd.DataFrame([
    {"etat": etat, "grande_region":region}
    for etat, region in etat_region.items()
])

print("total observation de la dimension region:", len(dim_region))
```

```python
# apercu du dataframe
print(dim_region.head(10))
```

```python
# verification
assert len(dim_region) == 27, f"resultat attendu:27, resulat obtenu: {len(dim_region)}"
assert dim_region["etat"].is_unique, "etats dupliques"
assert dim_region["grande_region"].nunique()==5, "pas 5 region"
assert dim_region.isna().sum().sum() == 0, " valeurs manquantes"

print(" controle des dimension")
print("total observation de la dimension region:", len(dim_region))
print("regions:", sorted(dim_region["grande_region"].unique()))
```

```python
# sauvegrde dans un fichier CSV
from pathlib import Path
Path("../data/processed/").mkdir(parents=True, exist_ok=True)
dim_region.to_csv("../data/processed/dim_region.csv", index=False, encoding="utf-8")
print("fichier CSV cree")
```
