
### EXPLORATION DES DONNEES DE Olist

```python
# importation des librairies
import pandas as pd
from pathlib import Path

from IPython.display import display
```

```python
data_dir=Path("../data/raw")
# creation de la liste des 9 fichiers csv
csv_files=[
        "olist_customers_dataset.csv",
    "olist_geolocation_dataset.csv",
    "olist_order_items_dataset.csv",
    "olist_order_payments_dataset.csv",
    "olist_order_reviews_dataset.csv",
    "olist_orders_dataset.csv",
    "olist_products_dataset.csv",
    "olist_sellers_dataset.csv",
    "product_category_name_translation.csv"
]

# Chargement de  tous les fichiers csv dans un dictionnaire de dataframe
dataframes = {}
print("Chargement des fichiers  csv Olist\n")

for fichier in csv_files:
    nom = fichier.replace("olist_", "").replace(".csv", "")
    chemin = data_dir / fichier
    data_file = pd.read_csv(chemin)
    dataframes[nom] = data_file

    nb_lignes = data_file.shape[0]
    nb_colonnes = data_file.shape[1]
    print(f"Fichier {nom} :   {nb_lignes} lignes, {nb_colonnes} colonnes")
```

```python
# apercu de chaque fichier
for nom, df in dataframes.items():
    print("colonne:", len(df.columns))
    print(list(df.columns))
    print("apercu\n")
    display(df.head(3))
```

### verification des anomalies

```python
# clients
customers= dataframes["customers_dataset"]
print("customers informations:\n")
print("unique customer_id :", customers["customer_id"].nunique())
print("unique customer_unique_id :", customers["customer_unique_id"].nunique())
# calcul du ratio
ratio =customers["customer_id"].nunique()/customers["customer_unique_id"].nunique()
print(f"Ratio : {ratio}")

# articles
items= dataframes["order_items_dataset"]
items_par_cmde=items.groupby("order_id").size()
print("\norder items:")
print("article par commande:",round(items_par_cmde.mean(),2))
print("commande ayant le maximum d'article: ", items_par_cmde.max())
print("commande avec un seul article: ",(items_par_cmde==1).sum())

# avis sur les commandes
reviews=dataframes["order_reviews_dataset"]
avis_par_cmde= reviews.groupby("order_id").size()
print("order reviews:\n")
print("total avis:", len(reviews))
print("commandes avec avis :", reviews["order_id"].nunique())
print("avis par commande moyen :",round(avis_par_cmde.mean(),2))
print("commande avec plusieurs avis: ", (avis_par_cmde>1).sum())

# paiements
paiements=dataframes["order_payments_dataset"]
paiements_par_cmde=paiments.groupby("order_id").size()
print("order payments:\n")
print("paiement par commande moyen: ", round(paiments_par_cmde.mean(),2))
print("commande avec plusieurs paiements:", (paiments_par_cmde>1).sum())

# geolocation
geolocations=dataframes["geolocation_dataset"]
print("geolocalisation:\n")
print("total observations:", len(geolocations))
print("unique zip code:",geolocations["geolocation_zip_code_prefix"].nunique())
doublons_moyen= len(geolocations)/geolocations["geolocation_zip_code_prefix"].nunique()
print("doublons moyen:", doublons_moyen)
```

### Analyse des statuts et des dates manquantes

```python
# commande
orders=dataframes["orders_dataset"]
print("repartition des statuts par commandes:")
print(orders["order_status"].value_counts())
print("total commandes :", len(orders))

# verification des dates manquantes
dates=[
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]
print("\ninformations sur les dates manquantes:")
for date in dates:
    nbre_date_manquante=orders[date].isna().sum()
    pourcentage_date_manquantes=nbre_date_manquante/len(orders)*100
    print(f"{date}: {pourcentage_date_manquantes:.2f}%")

# informations sur les commandes livrees
cmde_livree=orders[orders["order_status"]=="delivered"]
print("\ntotal commandes livrees: ", len(cmde_livree))
print("total commande sans date de livraison:", cmde_livree["order_delivered_customer_date"].isna().sum())
```

```python
# Bon pourcentage de commandes à un seul article
nb_cmd_avec_articles = len(items_par_cmde)
print("commandes avec articles :", nb_cmd_avec_articles)
print("pourcentage à un seul article :", round((items_par_cmde == 1).sum() / nb_cmd_avec_articles * 100, 2))

# 2-3. Nombres exacts de dates manquantes
print("sans date client :", orders["order_delivered_customer_date"].isna().sum())
print("sans date transporteur :", orders["order_delivered_carrier_date"].isna().sum())

# 5. Commandes sans aucun article
print("commandes sans article :", (~orders["order_id"].isin(items["order_id"])).sum())

# 6. Clients avec plusieurs villes
villes = customers.groupby("customer_unique_id")["customer_city"].nunique()
print("clients avec plusieurs villes :", (villes > 1).sum())

# 7. Avis multiples : combien de dates distinctes par commande ?
multi = reviews[reviews.duplicated("order_id", keep=False)]
print(multi.groupby("order_id")["review_answer_timestamp"].nunique().value_counts())
```

```python
cmd = orders.merge(customers, on="customer_id")
dim_client = (cmd.sort_values("order_purchase_timestamp")
    .drop_duplicates("customer_unique_id", keep="last")
    [["customer_unique_id", "customer_zip_code_prefix", "customer_city", "customer_state"]])
assert len(dim_client) == 96096
```

```python
sans_article = orders[~orders["order_id"].isin(items["order_id"])]
print(sans_article["order_status"].value_counts())

avec_article = items.merge(orders[["order_id", "order_status"]], on="order_id")
print(avec_article.groupby("order_status")["price"].agg(["count", "sum"]))
```

```python
assert items['order_id'].nunique() == 98666
assert reviews['order_id'].nunique() == 98673
assert customers['customer_unique_id'].nunique() == 96096
assert orders['order_delivered_customer_date'].isna().sum() == 2965
assert (~orders['order_id'].isin(items['order_id'])).sum() == 775
```

```python
STATUS_EXCLUS = ['canceled', 'unavailable']
lignes = items.merge(orders[['order_id', 'order_status']], on='order_id')
valides = lignes[~lignes['order_status'].isin(STATUS_EXCLUS)]
assert abs(valides['price'].sum() - 13494400.74) < 0.01
```
