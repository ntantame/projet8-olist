```python
import pandas as pd

sellers = pd.read_csv('../data/raw/olist_sellers_dataset.csv', dtype={'seller_zip_code_prefix': str})
print(sellers.shape)
sellers.head()
```

```python
# construire et contrôler dim_vendeur

dim_vendeur = sellers[['seller_id', 'seller_city', 'seller_zip_code_prefix']].copy()
dim_vendeur.columns = ['seller_id', 'ville', 'code_postal_prefixe']
assert len(dim_vendeur) == 3095
assert dim_vendeur['seller_id'].is_unique
assert dim_vendeur['code_postal_prefixe'].str.len().eq(5).all()
assert dim_vendeur.isna().sum().sum() == 0
print("Contrôles DimVendeur OK")
dim_vendeur.head()
```
